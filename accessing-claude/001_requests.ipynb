```python
# Load env variables
from dotenv import load_dotenv
load_dotenv()
```

```python
# Create an API Client
from anthropic import Anthropic

client = Anthropic()
model = 'claude-sonnet-4-6'
```

```python
def append_message(messages: list, role: str, text: str):
    message = {
        'role': role,
        'content': text
    }
    messages.append(message)

def chat(messages: list, *, temperature: float = 0.0):
    message = client.messages.create(
            model=model,
            max_tokens=1000,
            messages=messages,
            temperature=temperature
        )
    return message.content[0].text 
```

```python
# Make a request

messages = []

append_message(messages, 'user', 'What is Quantum Computing?')
first = chat(messages)
append_message(messages, 'assistant', first)

append_message(messages, 'user', 'Give another response')
second = chat(messages, temperature= 0.5)
append_message(messages, 'assistant', second)

```

```python
print(f'First answer:\n{first}\n ---------------------')
print(f'Second answer:\n{second}\n ----------------------')
```

```python
new_messages = []
append_message(new_messages, 'user', 'Write a 1 sentence description of a fake database')

stream = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=new_messages,
        stream=True
    )

for event in stream:
    print(event)
```

```python
with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=new_messages
) as stream:
    for text in stream.text_stream:
        print(text, end='')

stream.get_final_message()
```
